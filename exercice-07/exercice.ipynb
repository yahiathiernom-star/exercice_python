```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
```

```python
passagers_moins_18_ans = (titanic["age"] < 18).sum()
```

```python
billets_plus_chers = titanic.nlargest(5, "fare")[
    ["fare", "class", "embark_town"]
]
print(billets_plus_chers)
```

```python
part_femmes_premiere_classe = (
    (titanic["class"] == "First") &
    (titanic["sex"] == "female")
).sum() / (titanic["class"] == "First").sum() * 100
```

```python
taux_survie_femmes_premiere_classe = titanic.loc[
    (titanic["class"] == "First") &
    (titanic["sex"] == "female"),
    "survived"
].mean() * 100
```

```python
taux_survie_hommes_troisieme_classe = titanic.loc[
    (titanic["class"] == "Third") &
    (titanic["sex"] == "male"),
    "survived"
].mean() * 100
```

```python
print("Passagers de moins de 18 ans :", passagers_moins_18_ans)

print(
    "Part des femmes en première classe :",
    round(part_femmes_premiere_classe, 2),
    "%"
)

print(
    "Survie des femmes de première classe :",
    round(taux_survie_femmes_premiere_classe, 2),
    "%"
)

print(
    "Survie des hommes de troisième classe :",
    round(taux_survie_hommes_troisieme_classe, 2),
    "%"
)
```
